```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Load TinyLlama (1.1B) - 4-bit quantized for 8GB RAM
model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)  # Quantized

# Define the original question
question = """
Roger has 5 apples. He buys 2 packages of apples. 
Each package has 5 apples. How many apples does he have now?
"""

# Generate response
inputs = tokenizer(question, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=50)
answer = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(answer)
```
